## Validacion interactiva
Esto es simplemente como esta en el archivo unet/training_setup/callbacks.py -> validate_model

```python
import numpy as np
import torch
from scipy.ndimage import gaussian_filter
from report_guided_annotation import extract_lesion_candidates
from picai_eval import evaluate
from tqdm import tqdm
```

```python
from piccai_challenge.picai_baseline.src.picai_baseline.unet.training_setup.neural_networks.unetr import UNETR
device = 'cuda'
unetr = UNETR(3, 2, (32, 256, 256)).to(device)
model = unetr
```

```python
all_valid_preds, all_valid_labels = [], []
stop_at = 10
count = 0
for valid_data in valid_gen:
        try:
            valid_images = valid_data['data'].to(device)
            print('Shape first valid images', valid_images.shape)
            valid_labels = valid_data['seg']
            print('Shape valid labels:', valid_labels.shape)
        except Exception:
            valid_images = torch.from_numpy(valid_data['data']).to(device)
            print('Shape first valid images', valid_images.shape)
            valid_labels = torch.from_numpy(valid_data['seg'])
            print('Shape valid labels:', valid_labels.shape)

        # test-time augmentation
        valid_images = [valid_images, torch.flip(valid_images, [4]).to(device)] # El 4 es for la dimension que se flipea. 
        #print('VI after test time augmentation', valid_images)

        # aggregate all validation predictions
        # gaussian blur to counteract checkerboard artifacts in
        # predictions from the use of transposed conv. in the U-Net
        preds = [
            torch.sigmoid(model(x))[:, 1, ...].detach().cpu().numpy() # Se agarra el 1 pq es la prediccion de la lesion
            for x in valid_images
        ]
        print('Preds', len(preds))

        # revert horizontally flipped tta image
        preds[1] = np.flip(preds[1], [3])

        # gaussian blur to counteract checkerboard artifacts in
        # predictions from the use of transposed conv. in the U-Net
        all_valid_preds += [
            np.mean([
                gaussian_filter(x, sigma=1.5)
                for x in preds
            ], axis=0)
        ]
        print('all valid preds', len(all_valid_preds))
        all_valid_labels += [valid_labels.numpy()[:, 0, ...]]
        print('all valid labels', len(all_valid_labels))
        count += 1
        if stop_at == count:
             break
        
    
```

```python
final_det = np.concatenate([x for x in np.array(all_valid_preds)], axis=0)
final_pred = np.concatenate([x for x in np.array(all_valid_labels)], axis=0)
print(final_det.shape)
print(final_pred.shape)
```

```python
valid_metrics = evaluate(y_det=iter(final_det),
                         y_true=iter(final_pred),
                         y_det_postprocess_func=lambda pred: extract_lesion_candidates(pred)[0])
```
